```python
import os 
import pandas as pd
import numpy as np
```

```python
dir_base = "/home/mery/Desktop/procesar_results"

dir_db1000 = dir_base+"/db1000"
dir_db31341 = dir_base+"/db31341"
dir_dme = dir_base+"/dme_dnme"
```

```python
nsga2_df = pd.DataFrame(columns=["File", "NSGA2"])
nsga3_df = pd.DataFrame(columns=["File", "NSGA3"])
nhoed_df = pd.DataFrame(columns=["File", "NHOED"])
macoed_df = pd.DataFrame(columns=["File", "MACOED"])
unsga3_df = pd.DataFrame(columns=["File", "UNSGA3"])

file_name = "file_"
df_mean = pd.DataFrame(columns=["ID", "NSGA3", "NSGA2", "NHOED", "MACOED", "UNSGA3"])
df_stdr = pd.DataFrame(columns=["ID", "NSGA3", "NSGA2", "NHOED", "MACOED", "UNSGA3"])

for subdir in os.listdir(dir_dme):
    
    if not subdir.startswith("hip"):
        continue

    dir_hip = os.path.join(dir_dme, subdir)

    for archivo in os.listdir(dir_hip):
        # Verificar que el archivo tenga extensión .csv
    
        file_route = os.path.join(dir_hip, archivo)
            
        df_hip = pd.read_csv(file_route, header=None)
        if "nhoed" in str.lower(archivo):
            parts = archivo.split("_")[1].split(".")
            formatted_part = str(parts[0]).zfill(2)
            nhoed_df.loc[formatted_part] = {"File": file_name+formatted_part, "NHOED": df_hip[0][0]}
        if "macoed" in str.lower(archivo):
            parts = archivo.split("_")[1].split(".")
            formatted_part = str(parts[0]).zfill(2)
            macoed_df.loc[formatted_part] = {"File": file_name+formatted_part, "MACOED": df_hip[0][0]}
        if "nsgaii_" in str.lower(archivo):
            parts = archivo.split("_")[2].split(".")
            formatted_part = str(parts[0]).zfill(2)
            nsga2_df.loc[formatted_part] = {"File": file_name+formatted_part, "NSGA2": df_hip[0][0]}
        if "unsga3" in str.lower(archivo):
            parts = archivo.split("-")
            formatted_part = str(parts[3]).zfill(2)
            unsga3_df.loc[formatted_part] = {"File": file_name+formatted_part, "UNSGA3": df_hip[0][0]}
        if "nsgaiii" in str.lower(archivo):
            parts = archivo.split("_")[1].split(".")
            formatted_part = str(parts[0]).zfill(2)
            nsga3_df.loc[formatted_part] = {"File": file_name+formatted_part, "NSGA3": df_hip[0][0]}

    nsga3_df.sort_index()
    nsga2_df.sort_index()
    nhoed_df.sort_index()
    macoed_df.sort_index()
    unsga3_df.sort_index()
    df_final = nsga3_df
    df_final = pd.merge(df_final, nsga2_df, on="File", how="outer")
    df_final = pd.merge(df_final, nhoed_df, on="File", how="outer")
    df_final = pd.merge(df_final, macoed_df, on="File", how="outer")
    df_final = pd.merge(df_final, unsga3_df, on="File", how="outer")

    # Guardar el DataFrame final en un archivo CSV
    #output_file = dir_hip+f"/df_final_{subdir}.csv"
    #df_final.to_csv(output_file, index=False)
    dict_values = df_final.iloc[:, 1:].mean().to_dict()
    dict_values["ID"] = str(subdir.split("_")[1]).zfill(2)
    df_mean.loc[str(subdir.split("_")[1]).zfill(2)] = dict_values

    dict_values_stdr = df_final.iloc[:, 1:].std().to_dict()
    dict_values_stdr["ID"] = str(subdir.split("_")[1]).zfill(2)
    df_stdr.loc[str(subdir.split("_")[1]).zfill(2)] = dict_values_stdr
```
